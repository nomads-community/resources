# Drug resistance mutation prevalence (simple)
This notebook is designed to import and visualise point mutation data aggregated by `nomadic summarize`. The notebook can be run one cell at a time (Shift-Enter) or all together ('Run All' above).

```python
import pandas as pd
import plotly.graph_objects as go
from pathlib import Path
```

# Settings

```python
# Enter the full path to your workspace e.g. /home/user/workspaces/project_name/
# workspace_path = "../.."
workspace_path = "/home/dan/Desktop/nomadic_workspaces/MARS/"
```

```python
# You don't necessarily need to change any of the values below unless you want to change the default behaviour of the notebook

# Convert the workspace path to a Path object
workspace_path = Path(workspace_path)

# Check the folder is present
if not workspace_path.exists():
    raise FileNotFoundError(f"WARNING: Workspace not found at {workspace_path}")

#Define subfolders using the standard nomadic structure
summaries_path = workspace_path / "summaries" / workspace_path.name
master_csv_path = summaries_path / "metadata" / "master_metadata.csv"

# Decide whether you want the outputs to be saved and in which format
save_results = True
save_format = "svg"

# Define where the outputs will be saved
output_dir = Path.cwd() / "results" / workspace_path.name

if save_results:
    output_dir.mkdir(parents=True, exist_ok=True)
    print(f"All results will be saved to: {output_dir}")
```

# Prepare data

### Load and filter

```python
# Master metadata i.e.
master_df = pd.read_csv(master_csv_path)

# Load the variant data output by nomadic
aa_changes = pd.read_csv(summaries_path / "variants" / "prevalence.aa_changes.resistance.csv" )
```

# Plots

### Numbers

```python
# Define which gene you want to plot
gene = "dhfr"

# Plot the data
plot_df = aa_changes[aa_changes["gene"] == gene]
fig = go.Figure([
    go.Bar(x=plot_df["aa_change"], y=plot_df["n_mut"], name="Mutant", marker_color="red"),
    go.Bar(x=plot_df["aa_change"], y=plot_df["n_mixed"], name="Mixed", marker_color="orange"),
    go.Bar(x=plot_df["aa_change"], y=plot_df["n_wt"], name="WT", marker_color="lightgray"),
])
fig.update_layout(title=f"{gene} mutations",barmode="stack", xaxis_title="Mutation", yaxis_title="Number of samples")
fig.show()

# Save the figure if required
if save_results:
    fig.write_image(
        output_dir / f"{gene}_mutations.{save_format}"
    )   
```

### Prevalence

```python
# Define which gene you want to plot
gene = "dhfr"

# Plot the data
plot_df = aa_changes[aa_changes["gene"] == gene]
fig = go.Figure([
    go.Bar(x=plot_df["aa_change"], y=plot_df["prevalence"], marker_color="red"),
])
fig.update_layout(title=f"{gene} mutations",barmode="stack", xaxis_title="Mutation", yaxis_title="Mutation prevalence (%)")
fig.show()

# Save the figure if required
if save_results:
    fig.write_image(
        output_dir / f"{gene}_mutation_prevalence.{save_format}"
    )    
```
